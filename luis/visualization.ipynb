```python
import altair as al
```

```python
from altair.datasets import data

movies = data.movies()
```

```python
rated_movies = movies.dropna(subset=["Production Budget", "IMDB Rating"]).copy()
rated_movies = rated_movies[rated_movies["Production Budget"] > 0]

rated_movies["Budget group"] = "Up to $100M"
rated_movies.loc[rated_movies["Production Budget"] > 100000000, "Budget group"] = "Over $100M"
budget_ratings = rated_movies.groupby("Budget group")["IMDB Rating"].mean().reset_index()

chart = al.Chart(budget_ratings).mark_bar().encode(
    x=al.X("Budget group:N", sort=["Up to $100M", "Over $100M"],
           axis=al.Axis(labelAngle=0)),
    y=al.Y("IMDB Rating:Q", title="Average IMDb rating",
           scale=al.Scale(domain=[6, 7], zero=False)),
    y2=al.datum(6)
).properties(
    title="More money budgets, better movies",
    width=360,
    height=260
)
chart
```
